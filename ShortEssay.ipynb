```python
from bs4 import BeautifulSoup
import requests
from openpyxl import Workbook

wb = Workbook()
ws = wb.active

title = ["類別","區域","縣市","學校名稱","年級","科別","班級","作者","指導老師","作品標題","名次","連結","梯次"]
ws.append(title)

lst = [1110315,1101015,1100315,1091015,1090325,1081031,1080331]
#area = 10 #Taipei 1-19

for area in range(1,20):
    for batch in lst:

        url = f"https://www.shs.edu.tw/index.php?p=search&act=do_search&s_contest_number={batch}&s_area={area}&s_record=1&s_essay_cat=20&orderby=essay_name&order="
        response = requests.get(url)
        soup = BeautifulSoup(response.text, "html.parser")
        print(response)
        
        # result = soup.find_all("content")

        # print(soup.prettify())

        tag_name = 'table.data_table tr:nth-child(2)'
        articles = soup.select(tag_name)
        if articles == []:
            continue
        
        # print(articles)
        
        
        tag_name = 'table.data_table tr'
        articles = soup.select(tag_name)

        # articles[1]
        n=0
        row=[]
        for i in articles:
            if n==0:
                n+=1
                continue
            else:
                n+=1
                link = i.select("a")[0]

                row.append(i.text)

                # print(i.text)

                # print(link['href'])
                # print(i['href'])
        for i in row:
                tt = i.replace("\n","=").strip().split("=")
                tt.append(link['href'])
                tt.append(batch)
                tt1 = []
                for i in range(1,len(tt)):
                    if i == 10 or i == 12 or i==14:
                        continue
                    else:
                        tt1.append(tt[i])
                tt = tt1
                # print(tt[0])
                if tt[1] == "":
                    continue
                else:
                    ws.append(tt)
        space = [""]
        ws.append(space)

wb.save("data.xlsx")

# titles = soup.find("table", class_="data_table")
# print(titles)
# for i in titles:
#     print(title.select_one("a").getText())
```
